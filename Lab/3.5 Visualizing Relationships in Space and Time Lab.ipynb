```python
import plotly.express as px
import pandas as pd

df = pd.read_csv('global_co2_emissions.csv')

df.head()
```

```python
df.describe()
```

```python
fig = px.line(df,
              x="year",
              y="co2",
              color="Name",
              hover_data=["Name", "year", "co2"])

fig.update_layout(title="Global Cement CO2 Emissions by Country over Time",
                  xaxis_title="Year",
                  yaxis_title="CO2 Emissions")
fig.show()
```

This is a time-series dataset of Global Co2 Emissions. This visual clearly shows the top contributing countries to Co2 emissions and the exploding presence of Co2 emissions in recent years.

```python
#filter the dataframe to include only the Top 25% Name and Description with the highest co2_per_capita
df_top_contributors = df[df['co2'] >= df['co2'].quantile(0.50)]
df_top_contributors = df_top_contributors.sort_values('year')

#create an animated scatter plot of the Co2 emissions of the Top 25% countries with the highest co2_per_capita over time 
fig = px.scatter(df_top_contributors,
                 x="population",
                 y="co2",
                 size="co2",
                 color="Name",
                 hover_name="Description",
                 animation_frame="year",
                 title="Population vs. Cement CO2 Emissions over Time",
                 labels={"population": "Population", "cement_co2": "Cement CO2 Emissions"}
                 )
# Adjust the height of the figure and set y-axis range for the min and max of the df_top_contributors 'co2' column. Adjust the x-axis to a logarithmic scale with a min and max value based on the df_top_contributors 'population' column.
fig.update_layout(height=600,
                  yaxis=dict(range=[df_top_contributors['co2'].min(), df_top_contributors['co2'].max()]),
                  xaxis=dict(range=[df_top_contributors['population'].min(), df_top_contributors['population'].max()]))

fig.show()
```

Here we can see the Top-Co2-Contributing Regions with a timescale animation. This shows how population directly contributes to Co2 generation among developed contries. (I am having a very difficult time wiht the dataSet).

```python
# can we remove the 'World' entry from the dataframe for the choropleth map?
df_top_contributors = df_top_contributors[df_top_contributors['Name'] != 'World']
fig = px.choropleth(   df_top_contributors, 
                       locations='iso_code',
                       color='share_global_co2',
                       hover_name='Name',
                       color_continuous_scale = px.colors.sequential.Plasma,
                       projection='natural earth',
                        animation_frame="year",

                       title="Global CO2 Emissions by Country")

#change color to be a scale from 0 to 1
fig.update_coloraxes(colorbar_title="Share of Global CO2 Emissions", cmin=df_top_contributors['share_global_co2'].min(), cmax=df_top_contributors['share_global_co2'].max())




fig.update_layout(height=500)  

# Display the plot
fig.show()
```

This visualization allows the user to adjust the year and see how the different regions of the world have contributed to the global share of Co2 production.
